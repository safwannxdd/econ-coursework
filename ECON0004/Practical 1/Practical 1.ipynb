```python
import pandas as pd
```

```python
df = pd.read_excel("practical1.xlsx", sheet_name="Sheet1")
```

```python
Y = df["lnw"]
X = df["edu"]
```

```python
import statsmodels.api as sm
```

```python
X = sm.add_constant(X)
```

```python
model = sm.OLS(Y, X).fit()
```

```python
print(model.summary())
```
